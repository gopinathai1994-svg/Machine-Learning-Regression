```python
import pandas as pd
dataset=pd.read_csv("50_Startups.csv")
dataset
```

```python
dataset = pd.get_dummies(dataset,dtype=int,drop_first =True)
dataset
```

```python
dataset.columns
```

```python
independent = dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
dependent = dataset[['Profit']]
dependent
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.3,random_state=40)
```

```python
# algorthm
from sklearn.ensemble import RandomForestRegressor
regressor = RandomForestRegressor(n_estimators = 100,random_state=0,criterion='poisson')
regressor.fit(x_train,y_train)
```

```python
y_pred = regressor.predict(x_test)
y_pred
```

```python
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)
r_score
```

```python
import pickle
filename ="final_model_random_forest.sav"
pickle.dump(regressor,open(filename,'wb'))
loaded_model = pickle.load(open('final_model_random_forest.sav','rb'))
user_inputspend = float(input('Enter the value of R&D Spend: '))
user_administration = float(input('Enter the value of Administration: '))
marketing=  float(input('Enter the value of Marketing:'))
State_Florida = float(input('Enter the value of Florida:'))
State_New= float(input('Enter the value of New York:'))
result = loaded_model.predict([[user_inputspend,user_administration,marketing,State_Florida,State_New]])
result
```
