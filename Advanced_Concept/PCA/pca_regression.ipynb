```python
import pandas as pd
from pca import (PCAUnSup, Linear, svm_linear, 
    svm_NL, Decision, random, selectk_regression)
import warnings;
warnings.filterwarnings('ignore')
```

```python
dataset = pd.read_csv('Wine.csv')
X = dataset.drop('Customer_Segment',axis=1)
y = dataset['Customer_Segment']
X
```

```python
y
```

```python
# Splitting the dataset into the Training set and Test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
# 1. Get PCA explained variance values
getPCA = PCAUnSup(X_train, X_test, 8)

# 2. Convert to DataFrame and assign column name
df_pca = pd.DataFrame(getPCA, columns=['Explained Variance Ratio'])

# Display result
df_pca
```

```python
acclin=[]
accsvml=[]
accsvmnl=[]
accdes=[]
accrf=[]

r2_lin=Linear(X_train,y_train,X_test,y_test)
acclin.append(r2_lin)
print(acclin)
r2_sl=svm_linear(X_train,y_train,X_test,y_test)    
accsvml.append(r2_sl)

r2_NL=svm_NL(X_train,y_train,X_test,y_test)
accsvmnl.append(r2_NL)

r2_d=Decision(X_train,y_train,X_test,y_test)
accdes.append(r2_d)

r2_r=random(X_train,y_train,X_test,y_test)
accrf.append(r2_r)
    
    
result = selectk_regression(acclin, accsvml, accsvmnl, accdes, accrf)
result
```
