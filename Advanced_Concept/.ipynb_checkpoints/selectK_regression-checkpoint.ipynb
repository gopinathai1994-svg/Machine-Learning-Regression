```python
import pandas as pd
from selectk_Regression import (
    selectkbest, split_scalar, Linear, svm_linear, 
    svm_NL, Decision, random, selectk_regression
)
import warnings;
warnings.filterwarnings('ignore')
```

```python
dataset1 = pd.read_csv("prep.csv")
df2 = pd.get_dummies(dataset1,dtype=int, drop_first=True)
df2
```

```python
indep_X = df2.drop('classification_yes', axis=1)
dep_Y = df2['classification_yes']
```

```python
kbest, selected_cols = selectkbest(indep_X, dep_Y, 7)
X_train, X_test, y_train, y_test = split_scalar(kbest, dep_Y)
```

```python
pd.DataFrame(kbest,columns=selected_cols)
```

```python
acclin=[]
accsvml=[]
accsvmnl=[]
accdes=[]
accrf=[]

r2_lin=Linear(X_train,y_train,X_test,y_test)
acclin.append(r2_lin)
print(acclin)
r2_sl=svm_linear(X_train,y_train,X_test,y_test)    
accsvml.append(r2_sl)

r2_NL=svm_NL(X_train,y_train,X_test,y_test)
accsvmnl.append(r2_NL)

r2_d=Decision(X_train,y_train,X_test,y_test)
accdes.append(r2_d)

r2_r=random(X_train,y_train,X_test,y_test)
accrf.append(r2_r)
    
    
result = selectk_regression(acclin, accsvml, accsvmnl, accdes, accrf)
result
```
